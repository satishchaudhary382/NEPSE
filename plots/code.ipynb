```
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scienceplots
from scipy import stats

plt.style.use(['science', 'ieee', 'no-latex'])

df = pd.read_csv("/content/final_data.txt", sep="\t")

df.columns = [
    'Date', 'ex_rate_npr_per_usd', 'exports_india_npr_mn',
    'exports_row_npr_mn', 'exports_total_npr_mn', 'gdp_npr_mn',
    'gdp_usd_mn', 'imports_india_npr_mn', 'imports_total_npr_mn',
    'imports_row_npr_mn', 'india_real_gdp_usd', 'ln_export_india',
    'ln_export_row', 'ln_import_india', 'ln_india_tbr', 'ln_import_row',
    'ln_india_gdp', 'ln_nepal_gdp', 'ln_reer', 'ln_reer_neg', 'ln_row_gdp',
    'ln_reer_pos', 'ln_row_tbr', 'ln_total_exports', 'ln_total_tbr',
    'ln_total_imports', 'nepal_real_gdp_usd', 'reer', 'world_gdp_usd',
    'year'
]

df['Date'] = pd.to_datetime(df['Date'])
df = df.set_index('Date')


```

```
des_data = df[['ln_total_tbr','ln_india_tbr','ln_row_tbr',
    'ln_reer',
    'ln_nepal_gdp','ln_india_gdp','ln_row_gdp']]
```

```
des_result = des_data.describe()
```

```
des_result.loc['Skewness'] = stats.skew(des_data)
des_result.loc['kurtosis'] = stats.kurtosis(des_data)
```

```
jb_stats = []
jb_pvalues = []
for col in des_result.columns:
    jb_stat,jb_p_value = stats.jarque_bera(des_data[col])
    jb_stats.append(jb_stat)
    jb_pvalues.append(jb_p_value)
```

```
des_result.loc['Jarque-Bera Statistic'] = jb_stats
des_result.loc['Jarque-Bera p-value'] = jb_pvalues
```

```
multiplier = multiplier.dropna().reset_index().iloc[:,2:]
```

```
import matplotlib.pyplot as plt
import pandas as pd

# Assume your DataFrame is already loaded as 'multiplier'
# It should have columns: Period, Multiplier_ln_reer(+), Multiplier_ln_reer(-),
#                         Asymmetry_plot with C.I, Upper_C.I, Lower_C.I

# If 'Period' is not a column, use the index. Here we assume a 'Period' column.
# If not, create one: multiplier['Period'] = range(len(multiplier))

fig, ax = plt.subplots(figsize=(5, 3))

# Plot the positive shock multiplier (red dashed)
ax.plot(multiplier['Multiplier_ln_reer(+)'],
        color='#E74C3C', linestyle='--', linewidth=1.5, label='Multiplier ln_REER(+)')

# Plot the negative shock multiplier (black dashed)
ax.plot(multiplier['Multiplier_ln_reer(-)'],
        color='black', linestyle='--', linewidth=1.5, label='Multiplier ln_REER(-)')

# Plot the asymmetry line (blue solid, thicker)
ax.plot(multiplier['Asymmetry_plot with C.I'],
        color='#2E86C1', linestyle='-', linewidth=2, label='Asymmetry (Positive - Negative)')

# Plot confidence intervals for asymmetry (grey dotted)
ax.plot(multiplier['Upper_C.I'],
        color='gray', linestyle=':', linewidth=1.5, label='Upper C.I')
ax.plot(multiplier['Lower_C.I'],
        color='gray', linestyle=':', linewidth=1.5, label='Lower C.I')

# Zero reference line
# ax.axhline(y=0, color='black', linestyle='-', linewidth=0.5)

# Labels and title
ax.set_xlabel('Horizon (Periods)', fontsize=12)
ax.set_ylabel('Multiplier', fontsize=12)
# ax.set_title('NARDL Dynamic Multipliers with Asymmetry and Confidence Intervals',
#              fontsize=14, fontweight='bold')

# Grid
ax.grid(True, linestyle='--', alpha=0.3)

# Remove top/right spines and ticks
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='x', which='both', top=False)
ax.tick_params(axis='y', which='both', right=False)

# Tick label size
ax.tick_params(axis='x', labelsize=11)
ax.tick_params(axis='y', labelsize=11)

# Legend (position best to avoid overlap)
ax.legend(loc='upper right', fontsize=5)

# Compact layout
fig.tight_layout(pad=1.0)

# Save (PDF for publication)
fig.savefig('dynamic_multipliers_asymmetry.pdf', dpi=300, bbox_inches='tight', pad_inches=0.1)

plt.show()
```
